```python
Q1 = df.area.quantile(0.25)
Q3 = df.area.quantile(0.75)
IQR = Q3 - Q1
print(f'IQR : {IQR}')

lower_bound = Q1 - 1.5*IQR
upper_bound = Q3 + 1.5*IQR
outliers = [x for x in df['area'] if x < lower_bound or x > upper_bound]

print(f"Lower Bound for outliers: {lower_bound}")
print(f"Upper Bound for outliers: {upper_bound}")
print(f"Identified Outliers: {outliers}")
```

```python
sample_size = 20
sample = df.sample(sample_size, random_state=42)
sample_mean = np.mean(sample['area'])
sample_std = np.std(sample['area'])
confidence_level = 0.95
confidence_level_value = 1.96

margin_of_error = confidence_level_value*(sample_std/np.sqrt(sample_size))
lower_limit = sample_mean - margin_of_error
upper_limit = sample_mean + margin_of_error

print(f'Confidence Interval for the Area at {confidence_level*100} confidence:')
print()

print(f'Population Mean : {df.area.mean()}')
print(f'Sample Mean : {sample_mean}')
print(f'Margin of Error : {margin_of_error}')
print(f'Lower Limit : {lower_limit}')
print(f'Upper Limit : {upper_limit}')

if lower_limit < sample_mean < upper_limit:
    print("\nAssumptions are True")
else:
    print("\nAssumptions are False")
```

```python
prices = df['price']
areas = df['area']

print(f'Prices Mean: {df.price.mean()}')
print(f'Mean of Areas: {df.area.mean()}')
print()

alpha = 0.05

t_stats, p_value = stats.ttest_ind(prices, areas)
print(f'T-Statistic : {t_stats}')
print(f'P-Value : {p_value}')
print(f'Significance Level : {alpha}')
print()

if p_value > alpha:
    print("Accepted the Null Hypothesis: True Assumption")
else:
    print("Accepted the Null Hypothesis: False Assumption")
```
